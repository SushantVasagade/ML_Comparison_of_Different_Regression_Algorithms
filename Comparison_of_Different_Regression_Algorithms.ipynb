## 1. Linear Regression

### Concept

Linear Regression establishes a linear relationship between the input variables and a continuous target variable.

The general equation is:

$$
Y = \beta_0 + \beta_1X_1 + \beta_2X_2 + \cdots + \beta_nX_n
$$

where:

- $Y$ = predicted target value
- $\beta_0$ = intercept
- $\beta_1, \beta_2, \ldots, \beta_n$ = model coefficients
- $X_1, X_2, \ldots, X_n$ = input features

### Advantages

- Simple and easy to understand
- Fast to train
- Easy to interpret
- Works well when the relationship between features and target is approximately linear

### Limitation

Linear Regression may perform poorly when the relationship between input variables and the target is strongly nonlinear.

## 2. Polynomial Regression

### Concept

Polynomial Regression extends Linear Regression by adding polynomial terms of the original features.

For example:

$$
Y = \beta_0 + \beta_1X + \beta_2X^2
$$

It can model nonlinear relationships.

### Advantages

- Captures nonlinear patterns
- Still uses Linear Regression internally
- Can model curved relationships between variables

### Limitation

A high polynomial degree can create too many features and may cause overfitting.

## 3. Decision Tree Regression

### Concept

Decision Tree Regression predicts a continuous target by recursively splitting the dataset into smaller groups based on feature values.

A simplified decision process can be represented as:

$$
X_j < threshold
$$

The observations are divided into different regions, and the prediction at a leaf node is generally based on the average target value of the observations reaching that leaf.

### Advantages

- Can model nonlinear relationships
- Does not require feature scaling
- Easy to understand conceptually
- Can handle interactions between features

### Limitation

A very deep Decision Tree can overfit the training data and may not generalize well to unseen data.

## 4. Random Forest Regression

### Concept

Random Forest Regression is an ensemble learning algorithm that combines multiple Decision Trees.

Each tree produces a prediction, and the final prediction is obtained by averaging the predictions from all the trees.

The final prediction can be represented as:

$$
\hat{Y} = \frac{1}{N}\sum_{i=1}^{N}\hat{Y}_i
$$

where:

- $\hat{Y}$ = final predicted value
- $N$ = number of Decision Trees
- $\hat{Y}_i$ = prediction produced by the $i$-th tree

### Advantages

- Handles nonlinear relationships
- Provides stable predictions
- Reduces the overfitting tendency of a single Decision Tree
- Works well with complex datasets
- Can capture feature interactions

### Limitation

- More computationally expensive than a single Decision Tree
- Less interpretable than Linear Regression
- Requires more memory when many trees are used

## 5. Support Vector Regression (SVR)

### Concept

Support Vector Regression is an extension of Support Vector Machines used for predicting continuous numerical values.

SVR attempts to find a function that keeps prediction errors within a specified tolerance margin.

A simplified linear SVR function can be represented as:

$$
f(X) = w^TX + b
$$

where:

- $w$ = weight vector
- $X$ = input feature vector
- $b$ = bias term

For nonlinear relationships, SVR can use kernel functions such as the Radial Basis Function (RBF) kernel.

### Advantages

- Can model nonlinear relationships
- Effective in high-dimensional feature spaces
- RBF kernel can capture complex patterns
- Can provide good performance on appropriately scaled datasets

### Limitation

- Sensitive to feature scaling
- Parameter tuning is important
- Can be computationally expensive for large datasets

# 📏 Regression Evaluation Metrics

Regression models are evaluated using different metrics to measure the difference between the actual and predicted values.

## 1. Mean Absolute Error (MAE)

### Concept

Mean Absolute Error measures the average absolute difference between the actual target values and the predicted values.

The formula is:

$$
MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
$$

where:

- $n$ = number of observations
- $y_i$ = actual value
- $\hat{y}_i$ = predicted value

### Interpretation

MAE represents the average magnitude of prediction error without considering its direction.

### Evaluation

**Lower MAE indicates better prediction performance.**

---

## 2. Mean Squared Error (MSE)

### Concept

Mean Squared Error measures the average of the squared differences between actual and predicted values.

The formula is:

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

where:

- $n$ = number of observations
- $y_i$ = actual value
- $\hat{y}_i$ = predicted value

### Interpretation

Since the errors are squared, larger prediction errors receive greater importance.

### Evaluation

**Lower MSE indicates better prediction performance.**

---

## 3. Root Mean Squared Error (RMSE)

### Concept

Root Mean Squared Error is the square root of Mean Squared Error.

The formula is:

$$
RMSE = \sqrt{MSE}
$$

or:

$$
RMSE = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2}
$$

where:

- $n$ = number of observations
- $y_i$ = actual value
- $\hat{y}_i$ = predicted value

### Interpretation

RMSE represents the typical magnitude of prediction error and is expressed in the same unit as the target variable.

### Evaluation

**Lower RMSE indicates better prediction performance.**

---

## 4. R² Score

### Concept

R² Score, also called the coefficient of determination, measures how much of the variation in the target variable is explained by the regression model.

The formula is:

$$
R^2 = 1-\frac{SS_{res}}{SS_{tot}}
$$

where:

- $SS_{res}$ = residual sum of squares
- $SS_{tot}$ = total sum of squares

The residual sum of squares is:

$$
SS_{res} = \sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

The total sum of squares is:

$$
SS_{tot} = \sum_{i=1}^{n}(y_i-\bar{y})^2
$$

where:

- $y_i$ = actual value
- $\hat{y}_i$ = predicted value
- $\bar{y}$ = mean of actual target values

### Interpretation

A higher R² value generally indicates that the model explains more variation in the target variable.

### Evaluation

**Higher R² generally indicates better model performance.**

---

## 📊 Metric Comparison

| Metric | Measures | Better Value |
|---|---|---|
| **MAE** | Average absolute prediction error | Lower ↓ |
| **MSE** | Average squared prediction error | Lower ↓ |
| **RMSE** | Typical prediction error | Lower ↓ |
| **R² Score** | Variance explained by the model | Higher ↑ |

### Summary

In this experiment, **MAE, MSE, and RMSE should generally be minimized**, while **R² Score should generally be maximized** when comparing regression models on the same dataset.

### Explanation

This cell creates a `StandardScaler`.

Standardization transforms features using:

$$
z = \frac{x-\mu}{\sigma}
$$

where:

- $x$ = original value
- $\mu$ = mean
- $\sigma$ = standard deviation
- $z$ = standardized value

Feature scaling is especially important for **SVR** because it uses distance and kernel calculations.

```python
# Cell 16: Create StandardScaler

scaler = StandardScaler()

print("StandardScaler created successfully.")
```

```python
# Cell 17: Scale Training and Testing Features

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed successfully.")
```

## STEP 19 — Create Linear Regression Model

### Explanation

This cell creates a **Linear Regression** model.

Linear Regression assumes an approximately linear relationship between the input features and the continuous target.

The general equation is:

$$
Y = \beta_0 + \beta_1X_1 + \beta_2X_2 + \cdots + \beta_nX_n
$$

```python
# Cell 19: Create Linear Regression Model

linear_model = LinearRegression()

print("Linear Regression model created successfully.")
```

## STEP 20 — Create Polynomial Regression Model

### Explanation

This cell creates a Polynomial Regression pipeline.

The pipeline performs three operations:

1. Generates polynomial features of degree 2.
2. Standardizes the generated features.
3. Applies Linear Regression.

Polynomial Regression can model nonlinear relationships.

The pipeline is used to keep the transformation and model together and to make the process suitable for cross-validation.

```python
# Cell 20: Create Polynomial Regression Pipeline

polynomial_model = Pipeline([
    (
        "poly_features",
        PolynomialFeatures(
            degree=2,
            include_bias=False
        )
    ),
    (
        "scaler",
        StandardScaler()
    ),
    (
        "regressor",
        LinearRegression()
    )
])

print("Polynomial Regression pipeline created successfully.")
```

## STEP 21 — Create Decision Tree Regression Model

### Explanation

This cell creates a **Decision Tree Regressor**.

A Decision Tree divides the dataset into smaller regions using feature thresholds and predicts a continuous value for each final region.

`max_depth=15` limits the maximum depth of the tree and helps control excessive model complexity.

Decision Trees do not require feature scaling.

```python
# Cell 21: Create Decision Tree Regression Model

decision_tree_model = DecisionTreeRegressor(
    random_state=42,
    max_depth=15
)

print("Decision Tree Regression model created successfully.")
```

## STEP 22 — Create Random Forest Regression Model

### Explanation

This cell creates a **Random Forest Regressor**.

Random Forest is an ensemble algorithm that combines multiple Decision Trees.

The final prediction is calculated by averaging the predictions of the individual trees:

$$
\hat{Y} = \frac{1}{N}\sum_{i=1}^{N}\hat{Y}_i
$$

`n_estimators=100` creates 100 trees.

`n_jobs=-1` allows parallel processing using available CPU cores.

```python
# Cell 22: Create Random Forest Regression Model

random_forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

print("Random Forest Regression model created successfully.")
```

## STEP 23 — Create Support Vector Regression Model

```python
# Cell 23: Create Support Vector Regression Model

svr_model = SVR(
    kernel="rbf",
    C=100,
    epsilon=0.1
)

print("Support Vector Regression model created successfully.")
```

### Explanation

## STEP 28 — Calculate Regression Performance Metrics

### Explanation

This cell evaluates all five regression models using four metrics:

### MAE

$$
MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
$$

**Lower MAE is better.**

### MSE

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

**Lower MSE is better.**

### RMSE

$$
RMSE = \sqrt{MSE}
$$

**Lower RMSE is better.**

### R² Score

$$
R^2 = 1-\frac{SS_{res}}{SS_{tot}}
$$

**Higher R² is generally better.**

The results are stored in a Pandas DataFrame for comparison.

```python
# Cell 28: Calculate Regression Performance Metrics

results = []

for model_name, y_pred in predictions.items():

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    results.append({
        "Model": model_name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="R2 Score",
    ascending=False
).reset_index(drop=True)

display(results_df.round(4))
```

## STEP 29 — Visualize Model Performance

### Explanation

This cell visually compares the performance of all five regression algorithms.

Two graphs are created:

1. **RMSE Comparison**
2. **R² Score Comparison**

For RMSE, a lower value indicates better performance.

For R² Score, a higher value generally indicates better performance.

Visualization makes it easier to compare the models.

```python
# Cell 29: Visualize Model Performance

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].bar(
    results_df["Model"],
    results_df["RMSE"]
)

axes[0].set_title("RMSE Comparison")
axes[0].set_xlabel("Regression Model")
axes[0].set_ylabel("RMSE")
axes[0].tick_params(
    axis="x",
    rotation=45
)

axes[1].bar(
    results_df["Model"],
    results_df["R2 Score"]
)

axes[1].set_title("R² Score Comparison")
axes[1].set_xlabel("Regression Model")
axes[1].set_ylabel("R² Score")
axes[1].tick_params(
    axis="x",
    rotation=45
)

plt.tight_layout()
plt.show()
```

## STEP 30 — 5-Fold Cross-Validation and Final Comparison

### Explanation

This cell performs **5-Fold Cross-Validation** for all five regression algorithms.

The dataset is divided into five folds. Each fold is used once as validation data while the remaining folds are used for training.

The following metrics are calculated:

- **Mean CV R²** — Higher is better
- **Mean CV RMSE** — Lower is better
- **Mean CV MAE** — Lower is better

Progress is displayed after each model so that the execution status can be monitored.

Finally, the cell displays:

1. Cross-validation results
2. Test-set results
3. Model with the highest test-set R² score

The model is selected based on the actual experimental results.

```python
# Cell 30: 5-Fold Cross-Validation and Final Comparison

models_for_cv = {
    "Linear Regression": LinearRegression(),

    "Polynomial Regression": Pipeline([
        (
            "poly_features",
            PolynomialFeatures(
                degree=2,
                include_bias=False
            )
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "regressor",
            LinearRegression()
        )
    ]),

    "Decision Tree": DecisionTreeRegressor(
        random_state=42,
        max_depth=15
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=1
    ),

    "SVR": Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "regressor",
            SVR(
                kernel="rbf",
                C=100,
                epsilon=0.1
            )
        )
    ])
}


# Create 5-Fold Cross-Validation
cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# Store Cross-Validation Results
cv_results = []


print("=" * 60)
print("STARTING 5-FOLD CROSS-VALIDATION")
print("=" * 60)


# Evaluate each model
for name, model in models_for_cv.items():

    print(f"\nRunning: {name}")
    print("-" * 60)

    scores = cross_validate(
        model,
        X,
        y,
        cv=cv,
        scoring={
            "R2": "r2",
            "RMSE": "neg_root_mean_squared_error",
            "MAE": "neg_mean_absolute_error"
        },
        n_jobs=1
    )

    mean_r2 = scores["test_R2"].mean()
    mean_rmse = -scores["test_RMSE"].mean()
    mean_mae = -scores["test_MAE"].mean()

    cv_results.append({
        "Model": name,
        "Mean CV R2": mean_r2,
        "Mean CV RMSE": mean_rmse,
        "Mean CV MAE": mean_mae
    })

    print(f"Mean CV R²   : {mean_r2:.4f}")
    print(f"Mean CV RMSE : {mean_rmse:.4f}")
    print(f"Mean CV MAE  : {mean_mae:.4f}")


# Convert results into DataFrame
cv_results_df = pd.DataFrame(cv_results)


# Sort by R² score
cv_results_df = cv_results_df.sort_values(
    by="Mean CV R2",
    ascending=False
).reset_index(drop=True)


# Display Cross-Validation Results
print("\n")
print("=" * 60)
print("5-FOLD CROSS-VALIDATION RESULTS")
print("=" * 60)

display(
    cv_results_df.round(4)
)


# Display Test Set Results
print("\n")
print("=" * 60)
print("TEST SET RESULTS")
print("=" * 60)

display(
    results_df.round(4)
)


# Find model with highest Test R²
best_test_model = results_df.loc[
    results_df["R2 Score"].idxmax(),
    "Model"
]

best_test_r2 = results_df["R2 Score"].max()


# Display Final Result
print("\n")
print("=" * 60)
print("FINAL RESULT")
print("=" * 60)

print(f"Model with highest Test Set R²: {best_test_model}")
print(f"Test Set R² Score: {best_test_r2:.4f}")

print("\nExperiment completed successfully.")
```
